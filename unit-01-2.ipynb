## 1 · Names and values

The line `count = 0` stores the value `0` under the name `count`. The `=` means *store*, not
*equals*: the right side is worked out first, using the values the names hold right now, and the
result is stored under the name on the left. So `count = count + 1` is not an impossible equation —
it means *take what `count` holds, add 1, and store the result back under `count`*.

That pattern is so common it has a shorthand: `count += 1` does exactly the same thing.

**Predict what this cell prints, then run it.**

## 3 · `for` loops

The following cell contains a `for` loop. A `for` loop takes a list and performs a set of
instructions once for each element of that list. Each time around, the **loop variable** — here
`value`, a name you choose — holds the current element.

The indentation is not decoration. The indented lines *are* the loop's instructions, and the first
line back at the left margin is where the loop ends. Python reads the structure of your code off
the indentation, which is why a misplaced indent is an error and not a style complaint.

**Predict how many lines this prints, and what they are.**

````python
draws = [0.42, 0.71, 0.02]
for value in draws:
    print(value)
````

## 4 · `if`: doing something only sometimes

An `if` runs its indented instructions only when its test is true. The test is a comparison —
`value < 0.5` is a question, *is this value below 0.5?*, and its answer is `True` or `False`.

Put an `if` inside a `for` and the counter no longer counts every element — it counts the elements
that pass the test. This is the shape at the center of chapter 1, and of most code you will read
this term: **go through everything, count what passes**.

**For the list `[0.42, 0.71, 0.02]`, which values are below 0.5? Predict the printed number, then
run.**

````python
draws = [0.42, 0.71, 0.02]
count = 0
for value in draws:
    if value < 0.5:
        count += 1
print(count)
````

## 5 · `elif` and `else`

Tests can chain. `elif` means *otherwise, if* — its test is only asked when every test above it
has already failed. `else` catches whatever is left. Exactly one branch of the chain runs for each
value, top to bottom, first test that passes wins.

**Predict all three printed lines.**

````python
for value in [0.42, 0.71, 0.02]:
    if value < 0.3:
        print(value, "low")
    elif value < 0.6:
        print(value, "middle")
    else:
        print(value, "high")
````

Look at 0.02. It is below 0.3 *and* below 0.6 — both tests would pass — but it prints only `low`,
because once `value < 0.3` passes, the rest of the chain is skipped. An `elif` is not a second
`if`: it only gets asked when the line above said no.

Chapter 1's `sequence_site_leaky` turns on exactly this. When you reread it after today, ask of
its `elif`: *which reads ever reach this line?*

## 6 · `def` and `return`

`def` wraps a block of code, gives it a name, and lists its inputs — the **parameters** in the
parentheses. Nothing inside the block runs when Python reads the definition; it runs when you
**call** the function by name. `return` hands the answer back to whoever called.

Below is section 4's loop, wrapped so it can be asked three different questions without being
rewritten. **Predict all three printed numbers.**

````python
def count_below(draws, cutoff):
    count = 0
    for value in draws:
        if value < cutoff:
            count += 1
    return count

print(count_below([0.42, 0.71, 0.02], 0.5))
print(count_below([0.42, 0.71, 0.02], 0.8))
print(count_below([0.9, 0.9, 0.9, 0.9], 0.5))
````

One block of code, three different answers — that is what parameters buy. You can also name the
inputs when you call, which makes a call read like a sentence and is how chapter 1 writes them:

```python
count_below(draws=[0.42, 0.71, 0.02], cutoff=0.5)
```

## 8 · Random numbers, and the seed

`np.random.default_rng(18)` builds a random-number generator, and the 18 is its **seed**. Each
call to `rng.random()` hands out the next number in the generator's sequence, always in the range
[0, 1) — 0 is possible, 1 never happens. The same seed always produces the same sequence. That is
not a limitation; it is the property that makes code with randomness in it checkable at all, and
it is why every chunk in chapter 1 sets a seed.

Run the cell. You cannot predict these three numbers — nobody can, that is the point — but write
them down, to three decimal places. You are about to need them.

````python
import numpy as np

rng = np.random.default_rng(18)
print(rng.random())
print(rng.random())
print(rng.random())
````

**Now predict: if you build the generator again with the same seed, what do the same three calls
print?**

````python
rng = np.random.default_rng(18)
print(rng.random())
print(rng.random())
print(rng.random())
````

Identical, to the last digit. One warning that will save you a confused afternoon: the generator
hands out its sequence in order and never rewinds. If you re-run a cell that *calls* `rng.random()`
without re-running the line that *creates* `rng`, you get the next numbers in the sequence, not
the same ones again. When randomness looks wrong in this course, the first thing to check is
where the generator was created — and to re-run from there.

## 9 · Read a whole block

Every part is now on the table. The function below is chapter 1's `sequence_site` with the
methylation story removed — same loop, same test, same counter, plainer names.

Running the next cell prints nothing: it is a definition, and nothing inside runs until it is
called.

````python
def count_hits(chance, n_tries, rng):
    hits = 0
    for _ in range(n_tries):
        if rng.random() < chance:
            hits += 1
    return hits
````

**Four predictions, written down before you run.** The cell below calls `count_hits` four times,
and each call builds a fresh generator with seed 18 — so the first three numbers it will draw are
the three you wrote down in section 8.

1. `count_hits(0.5, 3, ...)` — trace it by hand: for each of your three written-down draws, is it
   below 0.5? How many pass?
2. `count_hits(1.0, 80, ...)` — you have not seen these 80 draws. You do not need to. What must
   the answer be?
3. `count_hits(0.0, 80, ...)` — same question.
4. `count_hits(0.5, 0, ...)` — how many times does the loop run?

````python
print(count_hits(0.5, 3, np.random.default_rng(18)))
print(count_hits(1.0, 80, np.random.default_rng(18)))
print(count_hits(0.0, 80, np.random.default_rng(18)))
print(count_hits(0.5, 0, np.random.default_rng(18)))
````

The last three are **forced**: you can say what must come out without seeing a single draw, purely
from the parts. `rng.random()` returns a number in [0, 1), so it is *always* below 1.0 — every one
of the 80 passes — and *never* below 0.0 — none do. And `range(0)` produces no numbers at all, so
the loop's instructions never run and `hits` is still 0 from the line above.

That is a check worth stealing. When you build your own simulator, find the input where the answer
is forced, and test the code against it before you trust anything else it tells you.
