## 1 · A comparison on an array asks every element at once

Compare a single value and you get one True/False answer (lab 2). Compare an **array** and you get one answer *per element* — an array of True/False called a **mask**. **Predict all three printed lines.**

```python
import numpy as np

draws = np.array([0.42, 0.71, 0.02, 0.88, 0.35])
is_small = (draws < 0.5)
print(is_small)
print(is_small.sum())
print(is_small.mean())
```

`[True False True False True]`, then 3, then 0.6. The last two are the counting trick that replaces half your loops: summing a mask counts its Trues (True adds 1), and its mean is the *fraction* of Trues. `count_below` from lab 2 is one line now — but only if the line carries a **name**: `is_small` states the question the mask asked. Convention in this course: a mask without a name does not pass review.

## 2 · Filtering: the mask as a keeper

An array indexed by a mask keeps exactly the elements that answered True — same length masks, element by element. **Predict both lines.**

```python
temperatures = np.array([36.5, 39.1, 36.9, 38.4, 37.0])
is_fever = (temperatures > 38.0)
print(temperatures[is_fever])
print(temperatures[is_fever].mean())
```

`[39.1 38.4]` and their mean. The filtered result is a fresh, shorter array; the original is untouched. And one mask can filter *another* array of the same length — that is chapter 4's whole engine: build the mask on one column (does this simulated case match?), use it to keep rows of another (who produced those cases?).

## 3 · Combining masks: | and &

Chapter 3's band filter used `|` — *either*. Its partner `&` means *both*. They combine masks element by element, and the rule of the house is a comment naming what the combination computes. **Predict the two counts.**

```python
values = np.array([3, 18, 9, 25, 1, 14])
outside_band = (values < 5) | (values > 20)    # too small OR too large
in_band = (values >= 5) & (values <= 20)       # within both fences
print(outside_band.sum(), in_band.sum())
```

3 and 3 — and no value escapes both masks, which is a structural check you can run on any band: outside plus inside must equal the total. (Parentheses around each comparison are required, not stylistic — `&` binds tighter than `<`.)
