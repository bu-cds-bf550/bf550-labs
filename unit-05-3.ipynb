## 1 · Trace the two-drawer lookup

**By hand: what prints?** Track which drawer each loop pass opens.

```python
rates = {"day": {"arrivals": 30, "departures": 28},
         "night": {"arrivals": 6, "departures": 9}}
total = 0
for shift in rates:
    table = rates[shift]
    total += table["arrivals"]
print(total)
print(len(rates), len(rates["day"]))
```

36 — the loop opened both drawers and took only the arrivals folder from each — then `2 2`: `len` counts *keys at that level*, not everything inside.

## 2 · Which comparison happens on which scale?

A colleague's snippet mixes scales. **Before running: the printed comparison — is it computed on probabilities or on logs, and does the conclusion survive the translation?**

```python
import numpy as np

p_first = 0.004
p_second = 0.0032
log_first = np.log(p_first)
log_second = np.log(p_second)
print("probability scale:", p_first > p_second)
print("log scale:       ", log_first > log_second)
```

True and True — logs keep order, always, which is the entire license for comparing log-scores as if they were the probabilities. The translation never flips a winner; it only rescues the arithmetic.
