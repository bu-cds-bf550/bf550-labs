## 1 · A crowd of simulated cases

Ten thousand simulated mornings: each is either a rush day (30% of them) or a calm day, and the visitor count is drawn poisson around that kind of day's rate. Two arrays, same length, positions aligned: `kind[i]` and `visitors[i]` describe the same morning — the alignment is what makes masks transfer. Read, then run.

```python
import numpy as np

rng = np.random.default_rng(101)
kind_list = []
visitor_list = []
for _ in range(10000):
    if rng.random() < 0.3:
        kind_list.append("rush")
        visitor_list.append(rng.poisson(60))
    else:
        kind_list.append("calm")
        visitor_list.append(rng.poisson(35))
kind = np.array(kind_list)
visitors = np.array(visitor_list)
print("rush share:", (kind == "rush").mean())
```

0.3-ish, as built. The two aligned arrays are the labeled corpus of chapter 4, story removed.

## 2 · The filtering answer

This morning saw **exactly 52 visitors**, and the door counter does not record the kind of day. Chapter 4's move, construct by construct: mask the crowd to mornings that look like this one, then ask what fraction of *those* were rush days. **Predict roughly, then run: with 52 sitting between the two rates, will the answer be near 0, near 1, or genuinely in between?**

```python
looks_like_today = (visitors == 52)
print("matching mornings:", looks_like_today.sum())
share_rush = (kind[looks_like_today] == "rush").mean()
print("share of them that were rush days:", round(share_rush, 2))
```

Genuinely in between — 52 is reachable by a lucky calm day or a slow rush day, and the crowd's split *is* the answer: that share is a probability, computed by counting. You have now run Bayes' theorem without writing it.
