## 1 · Every draw method is a machine with dials

`rng.random()` makes numbers in [0, 1), flat. The other draw methods are the same idea with different shapes and dials. The way to meet one is always the same: **draw a bunch, look at them, locate the dials.** Predict nothing yet — run and read.

```python
import numpy as np

rng = np.random.default_rng(53)
print("poisson(4):     ", rng.poisson(4, size=10))
print("gamma(2, 3):    ", np.round(rng.gamma(2, 3, size=6), 2))
print("beta(2, 8):     ", np.round(rng.beta(2, 8, size=6), 3))
```

Three signatures. `poisson(4)`: whole numbers, hovering near its dial (4) — a **count** machine. `gamma(2, 3)`: positive decimals, average near 2 × 3 = 6, occasionally large — a **positive amount** machine (rates, waiting times). `beta(2, 8)`: decimals strictly between 0 and 1, leaning toward 2/(2+8) = 0.2 — a **fraction** machine. Which machine a story needs is decided by what *kind* of number the story's quantity is.

## 2 · argmin and argmax: the position of the extreme

Chapter 3 hunts for *which candidate world* fits best — and the construct for "which" is new. `np.argmax(values)` returns the **position** of the largest value, never the value itself; `np.argmin`, the smallest. The name of the receiving variable should say it holds a position.

**Predict both printed numbers.**

```python
scores = np.array([0.31, 0.88, 0.12, 0.75])
best_position = np.argmax(scores)
print(best_position)
print(scores[best_position])
```

1, then 0.88 — the position, then the value fetched *with* the position. Writing `best = np.argmax(scores)` and then treating `best` as a score is one of the field's favorite silent bugs; the fix is in the name.
