## 1 · Slices: a piece of a list

You can pull one value out of a list by position. A **slice** pulls out a range of them: `values[start:stop]` — starting at `start`, stopping *before* `stop`, the same convention as `range`. Leave `start` empty and it means "from the beginning"; leave `stop` empty, "to the end".

**Predict all four printed lines.**

```python
values = [10, 20, 30, 40, 50, 60]
print(values[1:4])
print(values[:2])
print(values[4:])
print(values[2:3])
```

Take note:`values[2:3]` returns a one-value **list**,`[30]`, not the value 30. Also ensure that you understand that stop returns the elements before `stop`, which is why there are three elements returned by `values[1:4]`

## 2 · enumerate: positions and values together

`enumerate` is commonly used inside for loops where it will allow you to access the index and value of each item you're iterating over like `zip` with a built-in counter.

**Predict every printed line.**

```python
temperatures = [36.5, 36.9, 38.4, 37.1]
for position, temperature in enumerate(temperatures):
    if temperature > 38.0:
        print("reading", position, "is high:", temperature)
```

The for loop above will print exactly one line: "reading 2 is high: 38.4". The 2 is referring to its position in the `temperatures` list (recall that python uses 0-based indexing). `enumerate` and `[ ]` use the same numbering.

## 3 · Accumulate, then summarize

A common programming patterm will be to **run a model many times, append each result to a list, then take or compute one number from the list at the end.** . In this code below, an exact prediction is impossible; *why?*

```python
import numpy as np

rng = np.random.default_rng(29)
results = []                     # 1: start empty
for _ in range(2000):
    heads = 0
    for _ in range(20):
        if rng.random() < 0.5:
            heads += 1
    results.append(heads)        # 2: collect, one per run
print(np.mean(results))          # 3: summarize, once, after
print(np.std(results))
```

Reasonable predictions might be near 10 for the mean and near 2.2 for the standard deviation since twenty fair flips average ten heads, but the draws themselves are random. However, `results.append` will run exactly 2,000 times, and the two `print` lines run once.

## 4 · Arrays

`np.array` turns a list into an **array**, and arithmetic operations act on every element at once.
The **reductions** `.mean()` and `.sum()` do in one call what your loops did over many by collapsing values
down to a single number. 
**Predict all three printed lines before running.**

```python
import numpy as np

values = [3, 5, 4, 8]
arr = np.array(values)
print(arr * 2)
print(arr.sum())
print(arr.mean())
```

You'll notice that arithmetic (e.g. multiplying every value by 2) is applied to every element in the array
whereas `.sum()` and `.mean()` are reductions that collapse the array down to a single computed value.

## 5 · A thousand experiments in one call

Another concept you'll need to understand is rng.binomial, which draws samples from a binomial distribution and
takes three arguments: the number of trials (n), the probability of success (p), and how many experiments to run (size).

```python
rng = np.random.default_rng(3)
draws = rng.binomial(20, 0.5, 1000)
print(draws[:10])
print(draws.mean())
```

The code above runs one thousand twenty-flip experiments at once. The first printed line
shows the first 10 results and the second prints the average of all results.
