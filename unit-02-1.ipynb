## 1 · Slices: a piece of a list

You can pull one value out of a list by position. A **slice** pulls out a run of them: `values[start:stop]` — starting at `start`, stopping *before* `stop`, the same convention as `range`. Leave `start` empty and it means "from the beginning"; leave `stop` empty, "to the end".

**Predict all four printed lines.**

```python
values = [10, 20, 30, 40, 50, 60]
print(values[1:4])
print(values[:2])
print(values[4:])
print(values[2:3])
```

The last one surprises most readers: `values[2:3]` is `[30]` — a one-value **list**, not the value 30. A slice is always a list, however short. And count the elements of `values[1:4]`: three, not four — stop means *before*.

## 2 · enumerate: positions and values together

Sometimes a loop needs to know *where* it is, not just *what* it holds. `enumerate` hands you both, one pair per pass — the position first, the value second, like `zip` with a built-in counter.

**Predict every printed line.**

```python
temperatures = [36.5, 36.9, 38.4, 37.1]
for position, temperature in enumerate(temperatures):
    if temperature > 38.0:
        print("reading", position, "is high:", temperature)
```

One line prints — reading 2 — and the position is the useful part: it says *which* reading to go back and question. Positions start at 0 here exactly as they do in indexing; `enumerate` and `[ ]` speak the same language.

## 3 · Accumulate, then summarize — the pattern, named

Half of chapter 2's code is one shape: **run the world many times, append each result to a list, then take one number from the list at the end.** You built it in lab 2 without a name; the name makes it readable at a glance. Spot the three parts — the empty list, the append inside the loop, the summary after it — then predict roughly what prints (an exact prediction is impossible; *why?*).

```python
import numpy as np

rng = np.random.default_rng(29)
results = []                     # 1: start empty
for _ in range(2000):
    heads = 0
    for _ in range(20):
        if rng.random() < 0.5:
            heads += 1
    results.append(heads)        # 2: collect, one per world
print(np.mean(results))          # 3: summarize, once, after
print(np.std(results))
```

Near 10 and near 2.2 — *roughly* predictable from the story (twenty fair flips average ten heads), never exactly, because the draws are random. The structural reading is what is checkable: the `results.append` line runs 2,000 times; the two `print` lines run once. Indentation is the whole argument.

## 4 · Arrays: the list, upgraded

The chapter's first arrival. `np.array` turns a list into an **array**; arithmetic then acts
on every element at once, and the **reductions** `.mean()` and `.sum()` collapse it to one
number — what your loops did, in one call. **Predict all three printed lines before running.**

```python
import numpy as np

values = [3, 5, 4, 8]
arr = np.array(values)
print(arr * 2)
print(arr.sum())
print(arr.mean())
```

No loop touched any of that. The shape to remember: *arithmetic spreads across the array;
a reduction collapses it.*
