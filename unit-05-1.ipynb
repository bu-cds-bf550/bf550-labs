## 1 · A dict files values under names

A list files values under **positions**; a **dictionary** files them under **keys** — names you choose. Curly braces build one; square brackets look one up, exactly like list indexing with a name in place of the number.

**Predict all three printed lines.**

```python
depth = {"sample_a": 41000, "sample_b": 55000, "sample_c": 38000}
print(depth["sample_b"])
depth["sample_d"] = 61000
print(len(depth))
print("sample_e" in depth)
```

55000, then 4 — assignment to a new key *adds* it, and `len` counts the entries — then `False`: `in` asks whether a key exists, the polite alternative to looking up a missing key (which stops the program cold). `len` works on anything with a length — lists, dicts, arrays — and replaces the counting loop you have written five times by now.

## 3 · Counting into a dict

The most common dict pattern in the field: **tallying** — keys appear as they are met, values count. Read the guard (`in`) carefully, then predict the printed dict.

```python
calls = ["benign", "pathogenic", "benign", "benign", "vus"]
tally = {}
for call in calls:
    if call in tally:
        tally[call] += 1
    else:
        tally[call] = 1
print(tally)
```

`{'benign': 3, 'pathogenic': 1, 'vus': 1}`. The `else` branch creates; the `if` branch increments — without the guard, the first `+= 1` on a new key stops the program (there is nothing to add to yet).
