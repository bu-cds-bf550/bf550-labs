## 2 · Which world is this code building?

Read the chunk. **Before running: how many numbers does `spreads` hold when the loop finishes, and is each one built from 100 or from 1,000 draws?**

```python
import numpy as np

rng = np.random.default_rng(47)
spreads = []
for depth in (10, 100, 1000):
    fractions = []
    for _ in range(200):
        heads = 0
        for _ in range(depth):
            if rng.random() < 0.5:
                heads += 1
        fractions.append(heads / depth)
    spreads.append(np.std(fractions))
print(np.round(spreads, 3))
```

Three numbers — one per depth — each summarizing 200 worlds of `depth` flips. Three loops deep, and each level has a name: the dial (depth), the repetition (200 worlds), the world itself (the flips). Chapter 1's depth sweep is exactly this shape; being able to say *which loop is which* is the whole reading.
